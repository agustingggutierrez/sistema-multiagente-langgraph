# Demo del Orquestador Multiagente con LangGraph

Este notebook demuestra el flujo de delegación del sistema multiagente.

El grafo utiliza un nodo Supervisor que decide dinámicamente qué especialista debe intervenir según el estado compartido.

Flujo esperado:

**START → Supervisor → Research → Supervisor → Analyst → Supervisor → Validation → END**

Durante la ejecución se muestran:

- consulta del usuario;
- resultado de investigación;
- resultado del análisis;
- validación;
- respuesta final.
